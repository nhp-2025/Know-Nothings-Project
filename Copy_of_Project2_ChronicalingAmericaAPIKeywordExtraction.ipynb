```python
import requests
import pandas as pd
import time
from tqdm.auto import tqdm

base_url = "https://chroniclingamerica.loc.gov/search/pages/results/"
states = ["New York", "Massachusetts", "Maryland"]

all_results = []

for state in states:
    print(f"Collecting results for {state}")
    for _ in tqdm(range(50)):  # Adjust for sample size
            params = {
            "location_state": state,
            "start_date": "1845-01-01",
            "end_date": "1855-12-31",
            "andtext": "Know Nothings",
            "rows": 20,
            "format": "json",
            "sort": "random",
            "page": 1
        }

response = requests.get(base_url, params=params)
if response.status_code == 200:
      items = response.json().get("items", [])
      for item in items:
                    item['queried_state'] = state  # Add the state to each result
                     all_results.extend(items)
        else:
                print(f"Error: Received status code {response.status_code} for state {state}")
        except requests.exceptions.RequestException as e:
            print(f"Request failed for state {state}: {e}")

        time.sleep(1)  # to slow down request

# Convert to DataFrame
df = pd.DataFrame(all_results)
print(f"Total articles collected: {len(df)}")
```

```python
import re
from nltk.corpus import stopwords

# Load stopwords
stop_words = set(stopwords.words('english'))

# Define cleaner function
def clean_text(ocr_eng):
    # Convert to string and lowercase
    text = str(ocr_eng).lower()
    # Remove punctuation and numbers
    text = re.sub(r'[^a-z\s]', '', text)
    # Remove extra spaces
    text = re.sub(r'\s+', ' ', text).strip()
    # Remove stopwords
    words = [word for word in text.split() if word not in stop_words]
    return ' '.join(words)

```

```python
df['clean_text'] = df['ocr_eng'].fillna('').apply(clean_text)
```

```python
keyword_themes = {
    'catholicism': ['catholic', 'papist', 'romanism', 'jesuit', 'popery', 'pope', 'confessional', 'nunnery'],
    'immigration': ['irish', 'foreigner', 'immigrant', 'immigration', 'alien', 'foreign influence', 'assimilation', 'emigrant','emigration'],
    'moral': ['drunk', 'crime','invasion', 'alcohol', 'alcoholics'],
    'politics': ['american party', 'ballot', 'secret vote', 'order', 'platform', 'oath'],
    'violence': ['riot', 'mob', 'burning', 'lynch', 'torch', 'intimidation'],
    'patriotism': ['constitution', 'freedom', 'liberty', 'native american', 'flag','native born citizen', 'native','native born citizen']
}
```

```python
keyword_to_theme = {}
for theme, keywords in keyword_themes.items():
    for keyword in keywords:
        keyword_to_theme[keyword] = theme
```

```python
def assign_theme(extracted_keywords):
    for keyword in extracted_keywords:
        if keyword in keyword_to_theme:
            return keyword_to_theme[keyword]
    return "Other"  # Or assign a default theme if no match is found
```

```python
df['article_theme'] = df['clean_text'].apply(lambda text: assign_theme(text.split()))
```
